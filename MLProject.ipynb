##  Variable Cible (Class) : Succès du Film

### Définition du Succès

Pour ce projet de **Data Mining**, nous définissons le **succès d'un film** selon une **approche multicritère** combinant trois dimensions essentielles :

#### Critère Combiné   
Un film est considéré comme un **SUCCÈS** si et seulement si :
```python
success = 1  SI  {
    1. Rating >= 7.0           (Qualité Critique)
    ET
    2. Votes > médiane         (Popularité/Engagement)
    ET
    3. ROI >= 2.5             (Rentabilité Commerciale)
}
```

Où **ROI (Return on Investment)** = `grossWorldWide / budget`

### Justification du Critère

| Dimension | Indicateur | Seuil | Justification |
|-----------|------------|-------|---------------|
|  Qualité Critique | Rating IMDb | ≥ 7.0 | Un film avec une note ≥ 7/10 est généralement reconnu comme "bon" par la communauté IMDb |
|  Popularité | Votes | > médiane | Le film a généré suffisamment d'attention et d'engagement du public |
| Rentabilité | ROI | ≥ 2.5x | Le film a généré au moins 2.5 fois son budget, standard de rentabilité dans l'industrie cinématographique |

### **Merging** all 'merged_movies_data_[year]'.csv from the Dataset from 1920 to 2025

````python
# Path to main dataset folder
dataset_path = "/content/Data"

all_files = []

# Loop through each year folder
for year_folder in os.listdir(dataset_path):
    year_path = os.path.join(dataset_path, year_folder)
    if os.path.isdir(year_path):
        # Only pick CSVs starting with 'merged_movies_data_'
        csv_files = glob(os.path.join(year_path, "merged_movies_data_*.csv"))
        all_files.extend(csv_files)

# Read and concatenate all CSVs
df_list = [pd.read_csv(f) for f in all_files]
merged_df = pd.concat(df_list, ignore_index=True)

print("Merged dataset shape:", merged_df.shape)
merged_df.head()
````

Checking for a lost year

````python
# Extract the year from each file path
years_in_files = sorted([int(f.split('/')[-1].split('_')[-1].split('.')[0]) for f in all_files])
print("Years found in files:", years_in_files)

expected_years = list(range(1920, 2026))

# Find missing years
missing_years = [y for y in expected_years if y not in years_in_files]

print("Missing years:", missing_years)
````
